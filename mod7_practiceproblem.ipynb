```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv('diabetes.csv')

display(df.sample(14))

df.isnull().sum()
```

```python

cols = ['Glucose', 'BloodPressure', 'BMI']

imputer = SimpleImputer(missing_values=np.nan,strategy='median')

imputer.fit(df[cols])

df[cols] = imputer.transform(df[cols])

df.sample(14)
```

```python
# akne outliers detection korte bole ni tai iqr z_score no need

# khali bolse imputation korte mind it

# ata manually kortesi  

imputer_glocose = SimpleImputer(missing_values=np.nan, strategy='median')

imputer_glocose.fit(df[['Glucose']])

df['Glucose'] = imputer_glocose.transform(df[['Glucose']])

# bloodpressure
imputer_bloodpressure = SimpleImputer(missing_values=np.nan, strategy='median')

imputer_bloodpressure.fit(df[['BloodPressure']])

df['BloodPressure'] = imputer_bloodpressure.transform(df[['BloodPressure']])

# BMI
imputer_bmi = SimpleImputer(missing_values=np.nan, strategy='median')

imputer_bmi.fit(df[['BMI']])

df['BMI'] = imputer_bmi.transform(df[['BMI']])
```

```python
def age_group(age):

    if age < 30:
        return 'Young'

    elif age <= 50:
        return 'Adult'

    else:
        
        return 'Senior'



df['Age_Group'] = df['Age'].apply(age_group)

df[['Age','Age_Group']].head()

```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[['Young','Adult','Senior']])

encoder.fit(df[['Age_Group']])

df['Encoded_Age_Group'] = encoder.transform(df[['Age_Group']])

df[['Age_Group','Encoded_Age_Group']].head(14)    
```

```python
df=pd.read_csv('tips.csv')
df.sample(12)
```

```python
# one hot encoding korte hbe
sex_ohe=OneHotEncoder(sparse_output=False).set_output(transform='pandas')
sex_ohe.fit(df[['sex']])

encoded_sex_df=sex_ohe.transform(df[['sex']])

display(encoded_sex_df) # aknei oder jonno encoding sesh

# main df e concat korbo
df=pd.concat([df,encoded_sex_df],axis=1)
df

```

```python
# one hot encoding korte hbe
smoker_ohe=OneHotEncoder(sparse_output=False).set_output(transform='pandas')
smoker_ohe.fit(df[['smoker']])

encoded_smoker_df=smoker_ohe.transform(df[['smoker']])

display(encoded_smoker_df) # aknei oder jonno encoding sesh

# main df e concat korbo
df=pd.concat([df,encoded_smoker_df],axis=1)
df
```

```python
def tip_catagory(tip):
  if tip<3:
    return 'Low'

  elif tip<=3 and tip<=6:
    return 'Medium'

  else:
    return 'High'

df['tip_category']=df['tip'].apply(tip_catagory)
df
```

```python
sns.histplot(df['tip_category'])

# Encoding order frequency দেখে দেওয়া হয় না।

# Histogram শুধু দেখায়:
#কোন category কতবার আছে

# Shortcut মনে রাখো
# Frequency ≠ Order

# বারবার বেশি আসছে মানে higher rank না।
```

```python
tips_encoder=OrdinalEncoder(categories=[['Low','Medium','High']])
tips_encoder.fit(df[['tip_category']])

# ai ketre mol transform kori main df e oi column tate jeta nia kaj kortesi encoding er

df['tip_catagory']=tips_encoder.transform(df[['tip_category']]).ravel()

df



```

sex and smoker are binary categorical features with only two categories, so Label Encoding is the best choice because it converts categories into numeric values efficiently without creating extra columns.

tip_category has a meaningful order (Low < Medium < High), so Ordinal Encoding is the best technique because it preserves the natural ranking between the categories.

```python
df=pd.read_csv('heart.csv')
display(df.sample(10))
df.isnull().sum()
```

```python
x = df.drop('target', axis=1)

y = df['target']

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
print(x_train.isnull().sum())

# akne ami chol ata nia kaj korsi emni

chol_imputer = SimpleImputer(strategy='median')

# fit transform gola ak line e kore felsi

x_train[['chol']] = chol_imputer.fit_transform(x_train[['chol']])

x_test[['chol']] = chol_imputer.transform(x_test[['chol']])

Q1 = x_train['chol'].quantile(0.25)

Q3 = x_train['chol'].quantile(0.75)

IQR = Q3 - Q1
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

x_train = x_train[
    (x_train['chol'] >= lower_bound) |
    (x_train['chol'] <= upper_bound)
]
```

```python
print(df['cp'].unique())
```

```python

# One Hot Encoding


cp_ohe = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
).set_output(transform='pandas')

cp_ohe.fit(x_train[['cp']])

encoded_cp_x_train = cp_ohe.transform(
    x_train[['cp']]
)

encoded_cp_x_test = cp_ohe.transform(
    x_test[['cp']]
)


# Concat Encoded Columns


x_train = pd.concat(
    [x_train, encoded_cp_x_train],
    axis=1
)

x_test = pd.concat(
    [x_test, encoded_cp_x_test],
    axis=1
)

# Ordinal Encoding

ordinal_encoder = OrdinalEncoder()

x_train[['slope']] = ordinal_encoder.fit_transform(
    x_train[['slope']]
)

x_test[['slope']] = ordinal_encoder.transform(
    x_test[['slope']]
)


# Drop Original Columns


x_train.drop('cp', axis=1, inplace=True)

x_test.drop('cp', axis=1, inplace=True)


# Final Shape

print(x_train.shape)

print(x_test.shape)
```

# **Apply:**

**Write a short conclusion on which encoding is best for this feature.**


```python
df=pd.read_csv('diabetes.csv')
df
```

```python
def bmi_catagory(BMI):
  if BMI<18.5:
    return 'Underweight'

  elif BMI>=18.5 and BMI<=24.9:
    return 'Normal'

  elif BMI>=25.0 and BMI<=29.9:
    return 'Overweight'
  else:
    return 'Obese'
df['BMI_Catagory']=df['BMI'].apply(bmi_catagory)

```

```python
df
```

```python
#ordianl encodding
bmi_encoddeing=OrdinalEncoder(categories=[['Underweight','Normal','Overweight','Obese']])
bmi_encoddeing.fit(df[['BMI_Catagory']])

df['BMI_Catagory']=bmi_encoddeing.transform(df[['BMI_Catagory']])
df
```

```python
# one hot encodding
bmi_ohe=OneHotEncoder(sparse_output=False).set_output(transform='pandas')
bmi_ohe.fit(df[['BMI_Catagory']])
encoded_bmi_catagory_df = bmi_ohe.transform(df[['BMI_Catagory']]) # encoded_bmi_catagory_df atar moddei ora bibbino segment e bag hoi gese
display(encoded_bmi_catagory_df) # aknei oder jonno encoding sesh

# concat
df=pd.concat([df,encoded_bmi_catagory_df],axis=1) # main e add na kore new dataset open korsi
df
```

```python
# label encodding
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

le.fit(df['BMI_Catagory'])

df['BMI_Catagory_level']=le.transform(df['BMI_Catagory']) # akta new col create kore sekane raklam

df
# Label Encoding = শুধু নামগুলোকে সংখ্যা দেয়
# Ordinal Encoding = order (ছোট থেকে বড়) ধরে সংখ্যা দেয়

```

“Ordinal Encoding is the best choice for BMI category because it preserves the natural order between categories without creating unnecessary columns. Label Encoding may mislead the model by implying wrong relationships, while One Hot Encoding ignores the inherent ordering.”

## **8. Employee Attrition Dataset**

**– Advanced Encoding Challenge**
**Dataset Link:**

IBM HR Analytics Employee Attrition Dataset
Use the IBM HR Analytics Employee Attrition Dataset dataset.
# **Tasks:**

Split the dataset into train and test sets.

Detects all categorical columns automatically.

**Apply:**

Ordinal Encoding for education level

One Hot Encoding for department

Label Encoding for target column

Handle unseen categories in test data.
Create a reusable preprocessing function.( Optional )

Display transformed train and test datasets.


```python
df=pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')

df.sample(10)
```

```python

# x and y

x = df.drop('Attrition', axis=1)

y = df['Attrition']

# train test split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

# categorical columns
# ----------------------------------
# One Hot Encoding
# Department
# ----------------------------------

dept_ohe = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
).set_output(transform='pandas')

dept_ohe.fit(x_train[['Department']])

encoded_dept_x_train = dept_ohe.transform(
    x_train[['Department']]
)

encoded_dept_x_test = dept_ohe.transform(
    x_test[['Department']]
)

# ----------------------------------
# Concat Encoded Columns
# ----------------------------------

x_train = pd.concat(
    [x_train, encoded_dept_x_train],
    axis=1
)

x_test = pd.concat(
    [x_test, encoded_dept_x_test],
    axis=1
)

# ----------------------------------
# Ordinal Encoding
# EducationField
# ----------------------------------

ordinal_encoder = OrdinalEncoder()

x_train[['EducationField']] = ordinal_encoder.fit_transform(
    x_train[['EducationField']]
)

x_test[['EducationField']] = ordinal_encoder.transform(
    x_test[['EducationField']]
)

# ----------------------------------
# Drop Original Columns
# ----------------------------------

x_train.drop('Department', axis=1, inplace=True)

x_test.drop('Department', axis=1, inplace=True)

# ----------------------------------
# Label Encoding
# Attrition
# ----------------------------------

le = LabelEncoder()

y_train = le.fit_transform(y_train)

y_test = le.transform(y_test)

# ----------------------------------
# Final Shape
# ----------------------------------

print(x_train.shape)

print(x_test.shape)

# display

display(x_train.head())

display(x_test.head())
```
